```python
# ============================================================
# SPAM EMAIL CLASSIFICATION
# COMPLETE EDA
# Dataset: Kaggle Spam Email Dataset
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully! ✅")


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv("email_origin.csv")

print("\nDataset loaded successfully! ✅")


# ============================================================
# 3. FIRST 5 ROWS
# ============================================================

print("\n========== FIRST 5 ROWS ==========")

display(df.head())


# ============================================================
# 4. LAST 5 ROWS
# ============================================================

print("\n========== LAST 5 ROWS ==========")

display(df.tail())


# ============================================================
# 5. DATASET SHAPE
# ============================================================

print("\n========== DATASET SHAPE ==========")

print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])


# ============================================================
# 6. COLUMN NAMES
# ============================================================

print("\n========== COLUMN NAMES ==========")

print(df.columns.tolist())


# ============================================================
# 7. DATA TYPES
# ============================================================

print("\n========== DATA TYPES ==========")

print(df.dtypes)


# ============================================================
# 8. DATASET INFORMATION
# ============================================================

print("\n========== DATASET INFORMATION ==========")

df.info()


# ============================================================
# 9. MISSING VALUE ANALYSIS
# ============================================================

print("\n========== MISSING VALUE ANALYSIS ==========")

missing_values = df.isnull().sum()

display(
    missing_values.to_frame(
        "Missing Values"
    )
)

print(
    "\nTotal missing values:",
    df.isnull().sum().sum()
)


# ============================================================
# 10. MISSING VALUE PERCENTAGE
# ============================================================

print("\n========== MISSING VALUE PERCENTAGE ==========")

missing_percentage = (
    df.isnull().sum()
    / len(df)
    * 100
).round(2)

display(
    missing_percentage.to_frame(
        "Missing Percentage"
    )
)


# ============================================================
# 11. DUPLICATE ANALYSIS
# ============================================================

print("\n========== DUPLICATE ANALYSIS ==========")

duplicate_count = df.duplicated().sum()

print(
    "Duplicate rows:",
    duplicate_count
)

print(
    "Duplicate percentage:",
    round(
        duplicate_count / len(df) * 100,
        2
    ),
    "%"
)


# ============================================================
# 12. TARGET VARIABLE
# ============================================================

print("\n========== TARGET VARIABLE ==========")

print("Target variable: label")

print("Input feature: origin")

print(
    "\nProblem type: Binary Classification"
)

print(
    "Learning type: Supervised Learning"
)


# ============================================================
# 13. TARGET DISTRIBUTION
# ============================================================

print("\n========== TARGET DISTRIBUTION ==========")

target_counts = df["label"].value_counts()

display(
    target_counts.to_frame(
        "Number of Emails"
    )
)


# ============================================================
# 14. UNIQUE TARGET VALUES
# ============================================================

print("\n========== UNIQUE TARGET VALUES ==========")

print(
    df["label"].unique()
)


# ============================================================
# 15. TARGET PERCENTAGE
# ============================================================

print("\n========== TARGET PERCENTAGE ==========")

target_percentage = (
    df["label"]
    .value_counts(normalize=True)
    * 100
).round(2)

display(
    target_percentage.to_frame(
        "Percentage"
    )
)


# ============================================================
# 16. TARGET DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="label"
)

plt.title(
    "Spam vs Ham Email Distribution"
)

plt.xlabel(
    "Email Class"
)

plt.ylabel(
    "Number of Emails"
)

plt.tight_layout()

plt.show()


# ============================================================
# 17. CHECK MISSING EMAIL CONTENT
# ============================================================

print("\n========== EMAIL CONTENT CHECK ==========")

print(
    "Missing email contents:",
    df["origin"].isnull().sum()
)


# ============================================================
# 18. CREATE EMAIL LENGTH
# ============================================================

df["email_length"] = (
    df["origin"]
    .fillna("")
    .astype(str)
    .apply(len)
)

print(
    "\nEmail length feature created successfully! ✅"
)


# ============================================================
# 19. EMAIL LENGTH STATISTICS
# ============================================================

print("\n========== EMAIL LENGTH STATISTICS ==========")

display(
    df["email_length"].describe()
)


# ============================================================
# 20. AVERAGE EMAIL LENGTH BY CLASS
# ============================================================

print(
    "\n========== AVERAGE EMAIL LENGTH BY CLASS =========="
)

average_length = (
    df.groupby("label")["email_length"]
    .mean()
    .round(2)
)

display(
    average_length.to_frame(
        "Average Email Length"
    )
)


# ============================================================
# 21. MEDIAN EMAIL LENGTH BY CLASS
# ============================================================

print(
    "\n========== MEDIAN EMAIL LENGTH BY CLASS =========="
)

median_length = (
    df.groupby("label")["email_length"]
    .median()
)

display(
    median_length.to_frame(
        "Median Email Length"
    )
)


# ============================================================
# 22. OUTLIER ANALYSIS USING IQR
# ============================================================

print("\n========== OUTLIER ANALYSIS ==========")

Q1 = df["email_length"].quantile(0.25)

Q3 = df["email_length"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)

upper_bound = Q3 + (1.5 * IQR)

print("Q1          :", Q1)
print("Q3          :", Q3)
print("IQR         :", IQR)
print("Lower Bound :", lower_bound)
print("Upper Bound :", upper_bound)


# ============================================================
# 23. IDENTIFY OUTLIERS
# ============================================================

outliers = df[
    (df["email_length"] < lower_bound)
    |
    (df["email_length"] > upper_bound)
]

print(
    "\nPotential outliers:",
    len(outliers)
)

outlier_percentage = (
    len(outliers)
    / len(df)
    * 100
)

print(
    "Outlier percentage:",
    round(
        outlier_percentage,
        2
    ),
    "%"
)


# ============================================================
# 24. OUTLIERS BY CLASS
# ============================================================

print("\n========== OUTLIERS BY CLASS ==========")

display(
    outliers["label"]
    .value_counts()
    .to_frame(
        "Outlier Count"
    )
)


# ============================================================
# 25. OUTLIER BOXPLOT
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="label",
    y="email_length"
)

plt.title(
    "Email Length Distribution and Potential Outliers"
)

plt.xlabel(
    "Email Class"
)

plt.ylabel(
    "Email Length (Characters)"
)

plt.tight_layout()

plt.show()


# ============================================================
# 26. EMAIL LENGTH DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="email_length",
    hue="label",
    bins=50,
    kde=True
)

plt.title(
    "Email Length Distribution by Class"
)

plt.xlabel(
    "Email Length (Characters)"
)

plt.ylabel(
    "Number of Emails"
)

plt.tight_layout()

plt.show()


# ============================================================
# 27. SAMPLE HAM EMAILS
# ============================================================

print("\n========== SAMPLE HAM EMAILS ==========")

display(
    df[df["label"] == 0][
        [
            "label",
            "origin"
        ]
    ].head(3)
)


# ============================================================
# 28. SAMPLE SPAM EMAILS
# ============================================================

print("\n========== SAMPLE SPAM EMAILS ==========")

display(
    df[df["label"] == 1][
        [
            "label",
            "origin"
        ]
    ].head(3)
)


# ============================================================
# 29. RANDOM SAMPLE
# ============================================================

print("\n========== RANDOM SAMPLE ==========")

display(
    df[
        [
            "label",
            "origin"
        ]
    ].sample(
        min(5, len(df)),
        random_state=42
    )
)


# ============================================================
# 30. FINAL EDA SUMMARY
# ============================================================

print("\n")
print("=" * 65)

print(
    "                 FINAL EDA SUMMARY"
)

print("=" * 65)

print(
    "\nDataset size:",
    df.shape
)

print(
    "Input feature (X): origin"
)

print(
    "Target variable (y): label"
)

print(
    "Learning type: Supervised Learning"
)

print(
    "Problem type: Binary Classification"
)

print(
    "\nMissing values:",
    df.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print(
    "Potential outliers:",
    len(outliers)
)

print(
    "Outlier percentage:",
    round(
        outlier_percentage,
        2
    ),
    "%"
)

print(
    "\nTarget distribution:"
)

print(
    df["label"].value_counts()
)

print(
    "\nTarget percentage:"
)

print(
    target_percentage
)

print("\n")
print("=" * 65)

print(
    "                 EDA COMPLETED! ✅"
)

print("=" * 65)
```
