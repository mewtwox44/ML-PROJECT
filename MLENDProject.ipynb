```python
# ============================================================
# SPAM EMAIL CLASSIFICATION
# CEAS-08 DATASET
# Data Cleaning -> EDA -> Feature Extraction -> Train/Test
# -> Decision Tree + XGBoost -> Cross-Validation
# -> Evaluation & Comparison
# ============================================================

# =========================
# 1. INSTALL / IMPORT
# =========================

!pip install -q xgboost

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import warnings

warnings.filterwarnings("ignore")

from google.colab import files
from IPython.display import display

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)

from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.tree import DecisionTreeClassifier

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

print("=" * 70)
print("SPAM EMAIL CLASSIFICATION USING DECISION TREE AND XGBOOST")
print("=" * 70)


# =========================
# 2. UPLOAD DATASET
# =========================

print("\nPlease upload your CEAS_08.csv file")

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

print("\nUploaded file:", file_name)


# =========================
# 3. LOAD DATASET
# =========================

df = pd.read_csv(file_name)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())


# =========================
# 4. BASIC INFORMATION
# =========================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Info:")
df.info()


# =========================
# 5. STANDARDIZE COLUMN NAMES
# =========================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
)

print("\nStandardized columns:")
print(df.columns.tolist())


# =========================
# 6. CHECK REQUIRED COLUMNS
# =========================

required_columns = ["subject", "body", "label"]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns are missing: {missing_columns}\n"
        f"Available columns: {df.columns.tolist()}"
    )


# =========================
# 7. MISSING VALUE ANALYSIS
# =========================

print("\n" + "=" * 70)
print("MISSING VALUE ANALYSIS")
print("=" * 70)

print(df.isnull().sum())


# =========================
# 8. DUPLICATE ANALYSIS
# =========================

print("\n" + "=" * 70)
print("DUPLICATE ANALYSIS")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)


# =========================
# 9. HANDLE MISSING TEXT
# =========================

df["subject"] = df["subject"].fillna("")
df["body"] = df["body"].fillna("")

if "sender" in df.columns:
    df["sender"] = df["sender"].fillna("")

if "receiver" in df.columns:
    df["receiver"] = df["receiver"].fillna("")

if "date" in df.columns:
    df["date"] = df["date"].fillna("")

if "urls" in df.columns:
    df["urls"] = df["urls"].fillna("")


# =========================
# 10. CLEAN LABEL
# =========================

df["label"] = pd.to_numeric(
    df["label"],
    errors="coerce"
)

df = df.dropna(subset=["label"])

df["label"] = df["label"].astype(int)

print("\nLabel distribution:")
print(df["label"].value_counts())


# =========================
# 11. TEXT CLEANING FUNCTION
# =========================

def clean_text(text):

    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(
        r"<[^>]+>",
        " ",
        text
    )

    # Replace URLs
    text = re.sub(
        r"http\S+|www\S+",
        " URL ",
        text
    )

    # Replace email addresses
    text = re.sub(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
        " EMAIL ",
        text
    )

    # Replace numbers
    text = re.sub(
        r"\d+",
        " NUMBER ",
        text
    )

    # Keep letters and spaces
    text = re.sub(
        r"[^a-z\s]",
        " ",
        text
    )

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


# =========================
# 12. CREATE COMBINED EMAIL TEXT
# =========================

df["email_text"] = (
    df["subject"].astype(str)
    + " "
    + df["body"].astype(str)
)

df["clean_text"] = df["email_text"].apply(
    clean_text
)

print("\nCleaned text sample:")
display(
    df[["subject", "clean_text"]].head()
)


# ============================================================
# 13. EDA - EXPLORATORY DATA ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("EXPLORATORY DATA ANALYSIS")
print("=" * 70)


# -------------------------
# 13.1 Class Distribution
# -------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="label"
)

plt.title(
    "Spam vs Legitimate Email Distribution"
)

plt.xlabel("Email Class")
plt.ylabel("Number of Emails")

plt.xticks(
    [0, 1],
    ["Legitimate", "Spam"]
)

plt.show()


# -------------------------
# 13.2 Class Percentage
# -------------------------

class_percentage = (
    df["label"]
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

print("\nClass Distribution (%):")

for label, percentage in class_percentage.items():

    class_name = (
        "Legitimate"
        if label == 0
        else "Spam"
    )

    print(
        f"{class_name}: "
        f"{percentage:.2f}%"
    )


# -------------------------
# 13.3 Email Length
# -------------------------

df["email_length"] = (
    df["clean_text"]
    .str.len()
)

print("\nEmail Length Statistics:")
print(
    df["email_length"].describe()
)


plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="email_length",
    hue="label",
    bins=50,
    kde=True
)

plt.title(
    "Email Length Distribution"
)

plt.xlabel("Email Length")
plt.ylabel("Frequency")

plt.show()


# -------------------------
# 13.4 Word Count
# -------------------------

df["word_count"] = (
    df["clean_text"]
    .str.split()
    .str.len()
)

print("\nWord Count Statistics:")
print(
    df["word_count"].describe()
)


plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="label",
    y="word_count"
)

plt.title(
    "Word Count: Spam vs Legitimate"
)

plt.xlabel("Email Class")
plt.ylabel("Number of Words")

plt.xticks(
    [0, 1],
    ["Legitimate", "Spam"]
)

plt.show()


# -------------------------
# 13.5 Subject Length
# -------------------------

df["subject_length"] = (
    df["subject"]
    .astype(str)
    .str.len()
)

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="subject_length",
    hue="label",
    bins=40,
    kde=True
)

plt.title(
    "Subject Length Distribution"
)

plt.xlabel("Subject Length")
plt.ylabel("Frequency")

plt.show()


# -------------------------
# 13.6 URL Count
# -------------------------

def count_urls(text):

    return len(
        re.findall(
            r"http\S+|www\S+",
            str(text)
        )
    )


df["url_count"] = (
    df["email_text"]
    .apply(count_urls)
)

print("\nURL Count Statistics:")
print(
    df["url_count"].describe()
)


plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="label",
    y="url_count"
)

plt.title(
    "URL Count: Spam vs Legitimate"
)

plt.xlabel("Email Class")
plt.ylabel("Number of URLs")

plt.xticks(
    [0, 1],
    ["Legitimate", "Spam"]
)

plt.show()


# -------------------------
# 13.7 EDA Summary
# -------------------------

print("\nEDA Summary")
print("-" * 50)

print(
    "Total emails:",
    len(df)
)

print(
    "Legitimate emails:",
    (df["label"] == 0).sum()
)

print(
    "Spam emails:",
    (df["label"] == 1).sum()
)

print(
    "Average email length:",
    round(df["email_length"].mean(), 2)
)

print(
    "Average word count:",
    round(df["word_count"].mean(), 2)
)

print(
    "Average URL count:",
    round(df["url_count"].mean(), 2)
)


# ============================================================
# 14. FEATURE EXTRACTION
# ============================================================

print("\n" + "=" * 70)
print("FEATURE EXTRACTION USING TF-IDF")
print("=" * 70)

X_text = df["clean_text"]
y = df["label"]


# ============================================================
# 15. TRAIN / TEST SPLIT
# ============================================================

X_train_text, X_test_text, y_train, y_test = train_test_split(

    X_text,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)

print(
    "\nTraining samples:",
    len(X_train_text)
)

print(
    "Testing samples:",
    len(X_test_text)
)


# ============================================================
# 16. TF-IDF
# ============================================================

tfidf = TfidfVectorizer(

    max_features=10000,

    stop_words="english",

    ngram_range=(1, 2),

    min_df=2
)


X_train = tfidf.fit_transform(
    X_train_text
)

X_test = tfidf.transform(
    X_test_text
)

print(
    "\nTF-IDF Training Shape:",
    X_train.shape
)

print(
    "TF-IDF Testing Shape:",
    X_test.shape
)


# ============================================================
# 17. DECISION TREE
# ============================================================

print("\n" + "=" * 70)
print("DECISION TREE MODEL")
print("=" * 70)

dt_model = DecisionTreeClassifier(

    max_depth=30,

    random_state=42,

    class_weight="balanced"
)

dt_model.fit(
    X_train,
    y_train
)

dt_pred = dt_model.predict(
    X_test
)


# ============================================================
# 18. DECISION TREE EVALUATION
# ============================================================

dt_accuracy = accuracy_score(
    y_test,
    dt_pred
)

dt_precision = precision_score(
    y_test,
    dt_pred,
    zero_division=0
)

dt_recall = recall_score(
    y_test,
    dt_pred,
    zero_division=0
)

dt_f1 = f1_score(
    y_test,
    dt_pred,
    zero_division=0
)

print("\nDecision Tree Results")
print("-" * 40)

print(
    f"Accuracy : {dt_accuracy:.4f}"
)

print(
    f"Precision: {dt_precision:.4f}"
)

print(
    f"Recall   : {dt_recall:.4f}"
)

print(
    f"F1 Score : {dt_f1:.4f}"
)


print("\nClassification Report:")
print(
    classification_report(
        y_test,
        dt_pred,
        target_names=[
            "Legitimate",
            "Spam"
        ],
        zero_division=0
    )
)


# ============================================================
# 19. DECISION TREE CONFUSION MATRIX
# ============================================================

cm_dt = confusion_matrix(
    y_test,
    dt_pred
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_dt,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Legitimate",
        "Spam"
    ],
    yticklabels=[
        "Legitimate",
        "Spam"
    ]
)

plt.title(
    "Decision Tree Confusion Matrix"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()


# ============================================================
# 20. XGBOOST
# ============================================================

print("\n" + "=" * 70)
print("XGBOOST MODEL")
print("=" * 70)

xgb_model = XGBClassifier(

    n_estimators=200,

    max_depth=6,

    learning_rate=0.1,

    subsample=0.8,

    colsample_bytree=0.8,

    objective="binary:logistic",

    eval_metric="logloss",

    random_state=42,

    n_jobs=-1
)

xgb_model.fit(
    X_train,
    y_train
)

xgb_pred = xgb_model.predict(
    X_test
)


# ============================================================
# 21. XGBOOST EVALUATION
# ============================================================

xgb_accuracy = accuracy_score(
    y_test,
    xgb_pred
)

xgb_precision = precision_score(
    y_test,
    xgb_pred,
    zero_division=0
)

xgb_recall = recall_score(
    y_test,
    xgb_pred,
    zero_division=0
)

xgb_f1 = f1_score(
    y_test,
    xgb_pred,
    zero_division=0
)

print("\nXGBoost Results")
print("-" * 40)

print(
    f"Accuracy : {xgb_accuracy:.4f}"
)

print(
    f"Precision: {xgb_precision:.4f}"
)

print(
    f"Recall   : {xgb_recall:.4f}"
)

print(
    f"F1 Score : {xgb_f1:.4f}"
)


print("\nClassification Report:")

print(
    classification_report(
        y_test,
        xgb_pred,
        target_names=[
            "Legitimate",
            "Spam"
        ],
        zero_division=0
    )
)


# ============================================================
# 22. XGBOOST CONFUSION MATRIX
# ============================================================

cm_xgb = confusion_matrix(
    y_test,
    xgb_pred
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_xgb,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=[
        "Legitimate",
        "Spam"
    ],
    yticklabels=[
        "Legitimate",
        "Spam"
    ]
)

plt.title(
    "XGBoost Confusion Matrix"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()


# ============================================================
# 23. STRATIFIED 5-FOLD CROSS VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("5-FOLD CROSS-VALIDATION")
print("=" * 70)

cv = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=42
)


# ============================================================
# 24. DECISION TREE CROSS VALIDATION
# ============================================================

print("\nRunning Decision Tree Cross-Validation...")

dt_cv_results = cross_validate(

    dt_model,

    X_train,

    y_train,

    cv=cv,

    scoring=[
        "accuracy",
        "precision",
        "recall",
        "f1"
    ],

    n_jobs=-1
)


dt_cv_accuracy = (
    dt_cv_results[
        "test_accuracy"
    ].mean()
)

dt_cv_precision = (
    dt_cv_results[
        "test_precision"
    ].mean()
)

dt_cv_recall = (
    dt_cv_results[
        "test_recall"
    ].mean()
)

dt_cv_f1 = (
    dt_cv_results[
        "test_f1"
    ].mean()
)


print("\nDecision Tree 5-Fold CV")
print("-" * 40)

print(
    f"CV Accuracy : {dt_cv_accuracy:.4f}"
)

print(
    f"CV Precision: {dt_cv_precision:.4f}"
)

print(
    f"CV Recall   : {dt_cv_recall:.4f}"
)

print(
    f"CV F1 Score : {dt_cv_f1:.4f}"
)


# ============================================================
# 25. XGBOOST CROSS VALIDATION
# ============================================================

print("\nRunning XGBoost Cross-Validation...")

xgb_cv_results = cross_validate(

    xgb_model,

    X_train,

    y_train,

    cv=cv,

    scoring=[
        "accuracy",
        "precision",
        "recall",
        "f1"
    ],

    n_jobs=-1
)


xgb_cv_accuracy = (
    xgb_cv_results[
        "test_accuracy"
    ].mean()
)

xgb_cv_precision = (
    xgb_cv_results[
        "test_precision"
    ].mean()
)

xgb_cv_recall = (
    xgb_cv_results[
        "test_recall"
    ].mean()
)

xgb_cv_f1 = (
    xgb_cv_results[
        "test_f1"
    ].mean()
)


print("\nXGBoost 5-Fold CV")
print("-" * 40)

print(
    f"CV Accuracy : {xgb_cv_accuracy:.4f}"
)

print(
    f"CV Precision: {xgb_cv_precision:.4f}"
)

print(
    f"CV Recall   : {xgb_cv_recall:.4f}"
)

print(
    f"CV F1 Score : {xgb_cv_f1:.4f}"
)


# ============================================================
# 26. FINAL TEST SET COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("FINAL MODEL COMPARISON")
print("=" * 70)

comparison = pd.DataFrame({

    "Model": [
        "Decision Tree",
        "XGBoost"
    ],

    "Accuracy": [
        dt_accuracy,
        xgb_accuracy
    ],

    "Precision": [
        dt_precision,
        xgb_precision
    ],

    "Recall": [
        dt_recall,
        xgb_recall
    ],

    "F1 Score": [
        dt_f1,
        xgb_f1
    ]
})


display(
    comparison.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}"
    })
)


# ============================================================
# 27. CROSS-VALIDATION COMPARISON
# ============================================================

cv_comparison = pd.DataFrame({

    "Model": [
        "Decision Tree",
        "XGBoost"
    ],

    "CV Accuracy": [
        dt_cv_accuracy,
        xgb_cv_accuracy
    ],

    "CV Precision": [
        dt_cv_precision,
        xgb_cv_precision
    ],

    "CV Recall": [
        dt_cv_recall,
        xgb_cv_recall
    ],

    "CV F1 Score": [
        dt_cv_f1,
        xgb_cv_f1
    ]
})


print("\nCross-Validation Comparison:")

display(
    cv_comparison.style.format({
        "CV Accuracy": "{:.4f}",
        "CV Precision": "{:.4f}",
        "CV Recall": "{:.4f}",
        "CV F1 Score": "{:.4f}"
    })
)


# ============================================================
# 28. VISUAL MODEL COMPARISON
# ============================================================

plt.figure(figsize=(10, 6))

comparison_plot = comparison.set_index(
    "Model"
)

comparison_plot.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title(
    "Decision Tree vs XGBoost"
)

plt.ylabel("Score")

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=0
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 29. CROSS-VALIDATION SCORE VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

cv_plot = cv_comparison.set_index(
    "Model"
)

cv_plot.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title(
    "5-Fold Cross-Validation Comparison"
)

plt.ylabel("Mean CV Score")

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=0
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 30. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("PROJECT PIPELINE COMPLETED")
print("=" * 70)

print("""
CEAS-08 Dataset
       ↓
Data Cleaning
       ↓
Exploratory Data Analysis (EDA)
       ↓
TF-IDF Feature Extraction
       ↓
Train / Test Split
       ↓
Decision Tree + XGBoost
       ↓
5-Fold Stratified Cross-Validation
       ↓
Evaluation
       ↓
Model Comparison
""")

print("=" * 70)
print("FINAL TEST SET RESULTS")
print("=" * 70)

print(
    f"Decision Tree Accuracy : {dt_accuracy:.4f}"
)

print(
    f"Decision Tree F1 Score : {dt_f1:.4f}"
)

print(
    f"XGBoost Accuracy       : {xgb_accuracy:.4f}"
)

print(
    f"XGBoost F1 Score       : {xgb_f1:.4f}"
)

print("=" * 70)
print("NOTE: The values above are calculated from your uploaded dataset.")
print("=" * 70)
```
