```python
# ============================================================
# EDA - CEAS-08 Spam Email Dataset
# Project:
# Spam Email Classification Using Decision Trees and XGBoost
# with Cross-Validation
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. LOAD DATASET
# ============================================================

# CHANGE THIS PATH TO YOUR CSV FILE PATH
FILE_PATH = "CEAS_08.csv"

df = pd.read_csv(FILE_PATH)

print("=" * 60)
print("DATASET LOADED")
print("=" * 60)

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# 2. BASIC DATASET INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()


# ============================================================
# 3. STATISTICAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("STATISTICAL SUMMARY")
print("=" * 60)

print(df.describe(include="all"))


# ============================================================
# 4. MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUE ANALYSIS")
print("=" * 60)

missing_values = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_table = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percentage
})

print(missing_table)


# ============================================================
# 5. DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("DUPLICATE ANALYSIS")
print("=" * 60)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

print(
    "Percentage of duplicate rows:",
    round((duplicate_count / len(df)) * 100, 2),
    "%"
)


# ============================================================
# 6. LABEL DISTRIBUTION
# ============================================================

print("\n" + "=" * 60)
print("SPAM VS LEGITIMATE EMAILS")
print("=" * 60)

label_counts = df["label"].value_counts().sort_index()

print(label_counts)

print("\nPercentage distribution:")

print(
    (df["label"].value_counts(normalize=True) * 100)
    .round(2)
)


# ------------------------------------------------------------
# GRAPH 1: SPAM VS LEGITIMATE EMAILS
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

plt.bar(
    ["Legitimate", "Spam"],
    [
        label_counts.get(0, 0),
        label_counts.get(1, 0)
    ]
)

plt.title("Spam vs Legitimate Email Distribution")
plt.xlabel("Email Class")
plt.ylabel("Number of Emails")

plt.tight_layout()
plt.show()


# ============================================================
# 7. TEXT FEATURE ENGINEERING
# ============================================================

# Replace missing subject values
df["subject"] = (
    df["subject"]
    .fillna("")
    .astype(str)
)

# Replace missing body values
df["body"] = (
    df["body"]
    .fillna("")
    .astype(str)
)


# ------------------------------------------------------------
# SUBJECT LENGTH
# ------------------------------------------------------------

df["subject_length"] = (
    df["subject"].str.len()
)


# ------------------------------------------------------------
# BODY LENGTH
# ------------------------------------------------------------

df["body_length"] = (
    df["body"].str.len()
)


# ------------------------------------------------------------
# SUBJECT WORD COUNT
# ------------------------------------------------------------

df["subject_word_count"] = (
    df["subject"]
    .apply(lambda x: len(x.split()))
)


# ------------------------------------------------------------
# BODY WORD COUNT
# ------------------------------------------------------------

df["body_word_count"] = (
    df["body"]
    .apply(lambda x: len(x.split()))
)


print("\n" + "=" * 60)
print("TEXT FEATURE SUMMARY")
print("=" * 60)

print(
    df[
        [
            "subject_length",
            "body_length",
            "subject_word_count",
            "body_word_count"
        ]
    ].describe()
)


# ============================================================
# 8. URL ANALYSIS
# ============================================================

def count_urls(value):

    if pd.isna(value):
        return 0

    text = str(value).strip()

    if text == "":
        return 0

    # If URLs column already contains a number
    try:

        number = float(text)

        if np.isfinite(number):
            return int(number)

    except (ValueError, TypeError):

        pass

    # Detect URLs inside text
    urls = re.findall(
        r"(?:https?://|www\.)\S+",
        text
    )

    return len(urls)


df["url_count"] = (
    df["urls"]
    .apply(count_urls)
)


print("\n" + "=" * 60)
print("URL ANALYSIS")
print("=" * 60)

print(df["url_count"].describe())


# ============================================================
# 9. AVERAGE FEATURE VALUES BY CLASS
# ============================================================

numeric_features = [
    "subject_length",
    "body_length",
    "subject_word_count",
    "body_word_count",
    "url_count"
]

average_features = (
    df.groupby("label")[numeric_features]
    .mean()
    .round(2)
)

print("\n" + "=" * 60)
print("AVERAGE FEATURE VALUES BY EMAIL CLASS")
print("=" * 60)

print(average_features)


# ------------------------------------------------------------
# GRAPH 2: AVERAGE FEATURE VALUES
# ------------------------------------------------------------

plot_features = [
    "subject_length",
    "body_length",
    "subject_word_count",
    "body_word_count"
]

x = np.arange(len(plot_features))

width = 0.35

plt.figure(figsize=(11, 6))

plt.bar(
    x - width / 2,
    average_features.loc[0, plot_features],
    width,
    label="Legitimate"
)

plt.bar(
    x + width / 2,
    average_features.loc[1, plot_features],
    width,
    label="Spam"
)

plt.xticks(
    x,
    [
        "Subject Length",
        "Body Length",
        "Subject Word Count",
        "Body Word Count"
    ],
    rotation=20
)

plt.title("Average Email Features by Class")
plt.xlabel("Features")
plt.ylabel("Average Value")

plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# 10. SENDER ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("SENDER ANALYSIS")
print("=" * 60)

unique_senders = df["sender"].nunique()

print("Unique senders:", unique_senders)

top_senders = (
    df["sender"]
    .fillna("Unknown")
    .value_counts()
    .head(10)
)

print("\nTop 10 senders:")
print(top_senders)


# ------------------------------------------------------------
# GRAPH 3: TOP 10 SENDERS
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    range(len(top_senders)),
    top_senders.values
)

plt.xticks(
    range(len(top_senders)),
    top_senders.index,
    rotation=75,
    ha="right"
)

plt.title("Top 10 Senders")
plt.xlabel("Sender")
plt.ylabel("Number of Emails")

plt.tight_layout()
plt.show()


# ============================================================
# 11. RECEIVER ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("RECEIVER ANALYSIS")
print("=" * 60)

unique_receivers = df["receiver"].nunique()

print("Unique receivers:", unique_receivers)

top_receivers = (
    df["receiver"]
    .fillna("Unknown")
    .value_counts()
    .head(10)
)

print("\nTop 10 receivers:")
print(top_receivers)


# ------------------------------------------------------------
# GRAPH 4: TOP 10 RECEIVERS
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    range(len(top_receivers)),
    top_receivers.values
)

plt.xticks(
    range(len(top_receivers)),
    top_receivers.index,
    rotation=75,
    ha="right"
)

plt.title("Top 10 Receivers")
plt.xlabel("Receiver")
plt.ylabel("Number of Emails")

plt.tight_layout()
plt.show()


# ============================================================
# 12. DATE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("DATE ANALYSIS")
print("=" * 60)


# ------------------------------------------------------------
# CLEAN DATE COLUMN
# ------------------------------------------------------------

date_text = (
    df["date"]
    .fillna("")
    .astype(str)
    .str.strip()
)


# ------------------------------------------------------------
# CONVERT TO DATETIME
# utc=True handles different time zones
# ------------------------------------------------------------

df["date_clean"] = pd.to_datetime(
    date_text,
    errors="coerce",
    utc=True
)


# ------------------------------------------------------------
# COUNT VALID AND INVALID DATES
# ------------------------------------------------------------

valid_dates = df["date_clean"].notna().sum()

invalid_dates = df["date_clean"].isna().sum()

print("Valid dates:", valid_dates)

print("Invalid/missing dates:", invalid_dates)


# ------------------------------------------------------------
# DATE ANALYSIS
# ------------------------------------------------------------

if valid_dates > 0:

    # Keep only valid dates
    clean_dates = df.loc[
        df["date_clean"].notna(),
        "date_clean"
    ].copy()


    # --------------------------------------------------------
    # REMOVE EXTREME / INVALID YEAR VALUES
    # --------------------------------------------------------
    #
    # CEAS-08 contains some abnormal dates such as
    # 1980 and 2100.
    #
    # We use the reasonable email-data range
    # 2000 to 2020 for the year analysis.
    # --------------------------------------------------------

    reasonable_dates = clean_dates[
        (clean_dates.dt.year >= 2000) &
        (clean_dates.dt.year <= 2020)
    ].copy()


    # --------------------------------------------------------
    # COUNT DATE OUTLIERS
    # --------------------------------------------------------

    date_outliers = (
        len(clean_dates) -
        len(reasonable_dates)
    )

    print(
        "Date outliers:",
        date_outliers
    )


    # --------------------------------------------------------
    # EARLIEST AND LATEST REASONABLE DATE
    # --------------------------------------------------------

    if len(reasonable_dates) > 0:

        earliest_date = reasonable_dates.min()

        latest_date = reasonable_dates.max()

        print(
            "Earliest reasonable email:",
            earliest_date
        )

        print(
            "Latest reasonable email:",
            latest_date
        )


        # ====================================================
        # EMAILS BY YEAR
        # ====================================================

        year_counts = (
            reasonable_dates
            .dt.year
            .value_counts()
            .sort_index()
        )

        print("\nEmails by year:")
        print(year_counts)


        # ====================================================
        # GRAPH 5: EMAILS BY YEAR
        # ====================================================

        plt.figure(figsize=(10, 5))

        plt.plot(
            year_counts.index,
            year_counts.values,
            marker="o"
        )

        plt.title("Number of Emails by Year")

        plt.xlabel("Year")

        plt.ylabel("Number of Emails")

        plt.grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()

        plt.show()


    else:

        print(
            "No dates found in the reasonable year range."
        )


else:

    print("No valid dates found in the dataset.")

# ============================================================
# 13. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("CORRELATION ANALYSIS")
print("=" * 60)

correlation_features = [
    "label",
    "subject_length",
    "body_length",
    "subject_word_count",
    "body_word_count",
    "url_count"
]

correlation_matrix = (
    df[correlation_features]
    .corr()
    .round(2)
)

print(correlation_matrix)


# ============================================================
# 14. FINAL EDA SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("EDA COMPLETED")
print("=" * 60)

print("Total emails:", len(df))

print(
    "Total original columns:",
    7
)

print(
    "Spam emails:",
    (df["label"] == 1).sum()
)

print(
    "Legitimate emails:",
    (df["label"] == 0).sum()
)

print(
    "Duplicate rows:",
    duplicate_count
)

print(
    "Unique senders:",
    unique_senders
)

print(
    "Unique receivers:",
    unique_receivers
)

print(
    "Valid dates:",
    valid_dates
)

print(
    "Invalid/missing dates:",
    invalid_dates
)

print("\nEDA analysis completed successfully.")
```
